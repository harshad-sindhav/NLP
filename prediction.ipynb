```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle as p
```

```python
model=load_model('model.h5')
```

```python
with open ('label_gender.pkl','rb') as file :
    label_gender=p.load(file)
```

```python
with open ('one_hot_geo.pkl','rb') as file :
    one_hot_geo=p.load(file)
```

```python
with open ('Scaler.pkl','rb') as file :
    scaler=p.load(file)
```

```python
#  Sample data 
input_data = {

'CreditScore':600,

'Geography': 'France',

'Gender': 'Male',

'Age': 40,

'Tenure': 3,

'Balance':60000,

'NumOfProducts': 2,

'HasCrCard':1,

'IsActiveMember': 1,

'EstimatedSalary': 50000

}
```

```python
geo_encoded=one_hot_geo.transform([[input_data['Geography']]])
```

```python
geo_encoded.toarray()
```

```python
geo_encoded_df=pd.DataFrame(geo_encoded.toarray(),columns=one_hot_geo.get_feature_names_out(['Geography']))
```

```python
geo_encoded_df
```

```python
input_df=pd.DataFrame([input_data])
```

```python
input_df
```

```python
input_df['Gender']=label_gender.transform(input_df['Gender'])
```

```python
input_df
```

```python
input_df.drop('Geography',axis=1,inplace=True)
```

```python
input_df=pd.concat([input_df,geo_encoded_df],axis=1)
```

```python
input_df
```

```python
input_scaled=scaler.transform(input_df)
```

```python
input_scaled
```

```python
pred=model.predict(input_scaled)
```

```python
prediction_probability=pred[0][0]
if prediction_probability >0.5 :
    print('the customer is possibly leave')
else :
    print('customer will stay')
```
